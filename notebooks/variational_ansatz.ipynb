```python
import numpy as np
from qiskit import QuantumCircuit
from qiskit.circuit import Parameter
from qiskit.circuit.library import hamiltonian_variational_ansatz, evolved_operator_ansatz
from qiskit.quantum_info import SparsePauliOp
from qiskit.primitives import StatevectorEstimator
from scipy.optimize import minimize, OptimizeResult
from matplotlib import pyplot as plt
```

```python
hamiltonian = SparsePauliOp(
    [
        "IZZ",
        "ZIZ",
        "ZZI",
        "IXI",

    ],
    [
        0.5,
        0.5,
        0.5,
        0.5,
    ]
)
```

```python
ansatz = hamiltonian_variational_ansatz(hamiltonian)
num_params = ansatz.num_parameters
print(f"Number of parameters in the ansatz: {num_params}")
params = [Parameter(f"θ{i}") for i in range(num_params)]

ansatz.assign_parameters(params, inplace=True)
ansatz.draw("mpl")
```

```python
ansatz = evolved_operator_ansatz(hamiltonian)
num_params = ansatz.num_parameters
print(f"Number of parameters in the ansatz: {num_params}")
params = [Parameter(f"θ{i}") for i in range(num_params)]

ansatz.assign_parameters(params, inplace=True)
ansatz.draw("mpl")
```
